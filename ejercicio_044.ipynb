Implementa la clase `PersonajeJuegoV5` respetando los siguientes requerimientos:

1. **`__init__(self, nombre: str, vida: float)`**:
   - Inicializa las variables de instancia: `nombre` (str), `vida` (float), `inventario` (list vacía) y `activo` (bool = True).

2. **`evaluar_condicion(self, limite: float, factor: float) -> float`**:
   - Usa operadores aritméticos, lógicos y condicionales (`if / elif / else`).
   - Si `vida` >= limite y `activo` es True: retorna `vida` * factor.
   - Elif `vida` < limite y `vida` > 0: retorna `vida` + 10.0.
   - Else: retorna 0.0.

3. **`procesar_registros(self, datos: list) -> float`**:
   - Usa un ciclo `for` para recorrer `datos` (lista de floats/ints/None).
   - Ignora los elementos que sean `None`.
   - Acumula los valores válidos en `vida` y agrega cada valor a `inventario`.
   - Si `vida` supera los 500.0, desactiva la entidad (`activo = False`).
   - Retorna el valor final de `vida`.

4. **`acumular_hasta_objetivo(self, objetivo: float, paso: float) -> int`**:
   - Usa un ciclo `while` para incrementar `vida` en `paso` mientras `vida` sea estrictamente menor que `objetivo`.
   - Retorna la cantidad de iteraciones/pasos realizados.


```python
# Escribe tu solución aquí
class PersonajeJuegoV5:
    def __init__(self, nombre: str, vida: float):
            self.nombre = nombre
            self.vida = vida
            self.inventario = []
            self.activo = True
                


    def evaluar_condicion(self, limite: float, factor: float) -> float:
        if self.vida >= limite and self.activo == True:
             return self.vida * factor
        elif self.vida < limite and self.vida > 0:
                return self.vida + 10.0
        else:
             return 0.0
    

    def procesar_registros(self, datos: list) -> float:
        for valor in datos:
                if valor is None:
                   continue
                self.vida += valor
                self.inventario.append(valor)
                if self.vida > 500.0:
                   self.activo = False
        return self.vida
    

    def acumular_hasta_objetivo(self, objetivo: float, paso: float) -> int:
         contador = 0
         while self.vida < objetivo:
                self.vida += paso
                contador += 1
         return contador
    
```
