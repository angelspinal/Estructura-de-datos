## Objetivo

Implementar y comprender operaciones importantes de un Árbol Binario de Búsqueda.

Un BST organiza sus valores de la siguiente manera:

- Los valores menores que un nodo se almacenan en el subárbol izquierdo.
- Los valores mayores que un nodo se almacenan en el subárbol derecho.
- Cada nodo puede tener como máximo dos hijos.
- La raíz es el primer nodo del árbol.

Esta propiedad permite buscar, insertar y eliminar valores de manera eficiente cuando el árbol se encuentra balanceado.

## Estructura del árbol de prueba

Para realizar las pruebas se insertarán los siguientes valores:

```text
[10][20][25][30][35][40][45][50][60][70][80]
```

El árbol queda representado conceptualmente así:

```text
              50
            /    \
          30      70
         /  \    /  \
       20   40  60   80
      / \   / \
    10  25 35 45
```

- `50` es la raíz.
- `30` y `70` son hijos de `50`.
- `10`, `25`, `35`, `45`, `60` y `80` son hojas.
- Los valores menores se encuentran a la izquierda.
- Los valores mayores se encuentran a la derecha.

````python
from queue import Queue


class NodoArbol:
    def __init__(self, valor):
        self.valor = valor
        self.izquierdo = None
        self.derecho = None


class ArbolBinarioBusqueda:
    def __init__(self):
        self.raiz = None

    def insertar(self, valor):
        if self.raiz is None:
            self.raiz = NodoArbol(valor)
        else:
            self._insertar_recursivo(self.raiz, valor)

    def _insertar_recursivo(self, nodo_actual, valor):
        if valor < nodo_actual.valor:
            if nodo_actual.izquierdo is None:
                nodo_actual.izquierdo = NodoArbol(valor)
            else:
                self._insertar_recursivo(nodo_actual.izquierdo, valor)

        elif valor > nodo_actual.valor:
            if nodo_actual.derecho is None:
                nodo_actual.derecho = NodoArbol(valor)
            else:
                self._insertar_recursivo(nodo_actual.derecho, valor)

    def inorden(self):
        resultado = []
        self._inorden_recursivo(self.raiz, resultado)
        return resultado

    def _inorden_recursivo(self, nodo, resultado):
        if nodo is not None:
            self._inorden_recursivo(nodo.izquierdo, resultado)
            resultado.append(nodo.valor)
            self._inorden_recursivo(nodo.derecho, resultado)

    def eliminar(self, valor):
        self.raiz = self._eliminar_recursivo(self.raiz, valor)

    def _eliminar_recursivo(self, nodo, valor):
        if nodo is None:
            return None

        if valor < nodo.valor:
            nodo.izquierdo = self._eliminar_recursivo(
                nodo.izquierdo,
                valor
            )

        elif valor > nodo.valor:
            nodo.derecho = self._eliminar_recursivo(
                nodo.derecho,
                valor
            )

        else:
            if nodo.izquierdo is None and nodo.derecho is None:
                return None

            if nodo.izquierdo is None:
                return nodo.derecho

            if nodo.derecho is None:
                return nodo.izquierdo

            sucesor = self._minimo_nodo(nodo.derecho)
            nodo.valor = sucesor.valor
            nodo.derecho = self._eliminar_recursivo(
                nodo.derecho,
                sucesor.valor
            )

        return nodo

    def _minimo_nodo(self, nodo):
        actual = nodo

        while actual.izquierdo is not None:
            actual = actual.izquierdo

        return actual

    def recorrido_por_niveles(self):
        if self.raiz is None:
            return []

        resultado = []
        cola = Queue()
        cola.put(self.raiz)

        while not cola.empty():
            nodo_actual = cola.get()
            resultado.append(nodo_actual.valor)

            if nodo_actual.izquierdo is not None:
                cola.put(nodo_actual.izquierdo)

            if nodo_actual.derecho is not None:
                cola.put(nodo_actual.derecho)

        return resultado

    def sumar_elementos(self):
        return self._sumar_elementos_recursivo(self.raiz)

    def _sumar_elementos_recursivo(self, nodo):
        if nodo is None:
            return 0

        return (
            nodo.valor
            + self._sumar_elementos_recursivo(nodo.izquierdo)
            + self._sumar_elementos_recursivo(nodo.derecho)
        )

    def es_completo(self):
        return self._es_completo_recursivo(self.raiz)

    def _es_completo_recursivo(self, nodo):
        if nodo is None:
            return True

        if nodo.izquierdo is None and nodo.derecho is None:
            return True

        if nodo.izquierdo is None or nodo.derecho is None:
            return False

        return (
            self._es_completo_recursivo(nodo.izquierdo)
            and self._es_completo_recursivo(nodo.derecho)
        )

    def balancear_arbol(self):
        valores_ordenados = self.inorden()
        arbol_balanceado = ArbolBinarioBusqueda()

        arbol_balanceado.raiz = self._construir_balanceado(
            valores_ordenados,
            0,
            len(valores_ordenados) - 1
        )

        return arbol_balanceado

    def _construir_balanceado(self, valores, inicio, fin):
        if inicio > fin:
            return None

        medio = (inicio + fin) // 2
        nodo = NodoArbol(valores[medio])

        nodo.izquierdo = self._construir_balanceado(
            valores,
            inicio,
            medio - 1
        )

        nodo.derecho = self._construir_balanceado(
            valores,
            medio + 1,
            fin
        )

        return nodo

    def altura(self):
        return self._altura_recursiva(self.raiz)

    def _altura_recursiva(self, nodo):
        if nodo is None:
            return -1

        return 1 + max(
            self._altura_recursiva(nodo.izquierdo),
            self._altura_recursiva(nodo.derecho)
        )

    def mostrar_arbol(self):
        self._mostrar_nodo(self.raiz, "", True)

    def _mostrar_nodo(self, nodo, prefijo, es_izquierdo):
        if nodo is not None:
            siguiente_prefijo = prefijo + (
                "|   " if es_izquierdo else "    "
            )

            self._mostrar_nodo(
                nodo.derecho,
                siguiente_prefijo,
                False
            )

            simbolo = "\\-- " if es_izquierdo else "/-- "
            print(prefijo + simbolo + str(nodo.valor))

            siguiente_prefijo = prefijo + (
                "    " if es_izquierdo else "|   "
            )

            self._mostrar_nodo(
                nodo.izquierdo,
                siguiente_prefijo,
                True
            )
````

## Clases utilizadas

### Clase `NodoArbol`

La clase `NodoArbol` representa cada elemento individual del árbol.

Cada nodo tiene:

- `valor`: número almacenado.
- `izquierdo`: referencia al hijo izquierdo.
- `derecho`: referencia al hijo derecho.

Al crear un nodo, sus hijos comienzan en `None`, porque todavía no tiene conexiones.

### Clase `ArbolBinarioBusqueda`

La clase `ArbolBinarioBusqueda` administra la estructura completa.

Su atributo principal es `raiz`, que representa el primer nodo del árbol. También contiene métodos para insertar, recorrer, eliminar, sumar, validar y balancear los elementos.

# Ejercicio 1: Eliminar un nodo

Se implementa el método `eliminar(valor)` para remover un valor del BST.

La eliminación debe manejar tres casos:

1. **Nodo hoja:** no tiene hijos, por lo tanto se elimina directamente.
2. **Nodo con un hijo:** se reemplaza el nodo por su único hijo.
3. **Nodo con dos hijos:** se busca el sucesor inorden, que es el valor mínimo del subárbol derecho. Ese valor reemplaza al nodo eliminado y luego se elimina el sucesor de su posición original.

Esta estrategia conserva la propiedad del Árbol Binario de Búsqueda.

````python
def eliminar(self, valor):
    self.raiz = self._eliminar_recursivo(self.raiz, valor)

def _eliminar_recursivo(self, nodo, valor):
    if nodo is None:
        return None

    if valor < nodo.valor:
        nodo.izquierdo = self._eliminar_recursivo(
            nodo.izquierdo,
            valor
        )
    elif valor > nodo.valor:
        nodo.derecho = self._eliminar_recursivo(
            nodo.derecho,
            valor
        )
    else:
        if nodo.izquierdo is None and nodo.derecho is None:
            return None

        if nodo.izquierdo is None:
            return nodo.derecho

        if nodo.derecho is None:
            return nodo.izquierdo

        sucesor = self._minimo_nodo(nodo.derecho)
        nodo.valor = sucesor.valor
        nodo.derecho = self._eliminar_recursivo(
            nodo.derecho,
            sucesor.valor
        )

    return nodo
````

# Ejercicio 2: Recorrido por niveles BFS

BFS significa **Breadth First Search**, o recorrido en anchura.

El recorrido visita el árbol nivel por nivel:

1. Primero visita la raíz.
2. Después visita los hijos de la raíz.
3. Luego visita los nietos.
4. Continúa de izquierda a derecha.

Para realizarlo se usa una cola.

Una cola trabaja con la regla FIFO:

```text
First In, First Out
Primero en entrar, primero en salir
```

Cada nodo que sale de la cola agrega sus hijos al final de la misma.

````python
def recorrido_por_niveles(self):
    if self.raiz is None:
        return []

    resultado = []
    cola = Queue()
    cola.put(self.raiz)

    while not cola.empty():
        nodo_actual = cola.get()
        resultado.append(nodo_actual.valor)

        if nodo_actual.izquierdo is not None:
            cola.put(nodo_actual.izquierdo)

        if nodo_actual.derecho is not None:
            cola.put(nodo_actual.derecho)

    return resultado
````

# Ejercicio 3: Suma de elementos

Se implementa un método recursivo que suma todos los valores almacenados en el árbol sin modificar su estructura.

La lógica es:

\[
Suma(nodo) = nodo.valor + Suma(izquierdo) + Suma(derecho)
\]

El caso base ocurre cuando el nodo es `None`. En ese caso no existe valor que sumar, por lo tanto retorna `0`.

````python
def sumar_elementos(self):
    return self._sumar_elementos_recursivo(self.raiz)

def _sumar_elementos_recursivo(self, nodo):
    if nodo is None:
        return 0

    return (
        nodo.valor
        + self._sumar_elementos_recursivo(nodo.izquierdo)
        + self._sumar_elementos_recursivo(nodo.derecho)
    )
````

# Ejercicio 4: Validar nodos internos

El método `es_completo()` valida la condición solicitada en la tarea:

> Todos los nodos internos deben tener exactamente dos hijos.

Se consideran válidos:

- Un nodo vacío.
- Un nodo hoja, porque no es interno.
- Un nodo que tenga hijo izquierdo y derecho.

Se considera inválido un nodo que tenga solamente un hijo.

````python
def es_completo(self):
    return self._es_completo_recursivo(self.raiz)

def _es_completo_recursivo(self, nodo):
    if nodo is None:
        return True

    if nodo.izquierdo is None and nodo.derecho is None:
        return True

    if nodo.izquierdo is None or nodo.derecho is None:
        return False

    return (
        self._es_completo_recursivo(nodo.izquierdo)
        and self._es_completo_recursivo(nodo.derecho)
    )
````

# Ejercicio 5: Balancear el árbol

Un árbol puede quedar degenerado cuando los valores se insertan en orden. Por ejemplo:

```text
10
  \
   20
     \
      30
        \
         40
```

Este tipo de árbol se parece a una lista enlazada y sus operaciones pueden tardar \(O(n)\).

Para balancearlo se realizan dos pasos:

1. Se obtiene una lista ordenada mediante recorrido inorden.
2. Se selecciona el elemento central como raíz.
3. Se repite el proceso recursivamente con las mitades izquierda y derecha.

El método crea y retorna un árbol nuevo, por lo tanto no modifica el árbol original.

````python
def balancear_arbol(self):
    valores_ordenados = self.inorden()
    arbol_balanceado = ArbolBinarioBusqueda()

    arbol_balanceado.raiz = self._construir_balanceado(
        valores_ordenados,
        0,
        len(valores_ordenados) - 1
    )

    return arbol_balanceado

def _construir_balanceado(self, valores, inicio, fin):
    if inicio > fin:
        return None

    medio = (inicio + fin) // 2
    nodo = NodoArbol(valores[medio])

    nodo.izquierdo = self._construir_balanceado(
        valores,
        inicio,
        medio - 1
    )

    nodo.derecho = self._construir_balanceado(
        valores,
        medio + 1,
        fin
    )

    return nodo

def altura(self, nodo=None):
    if nodo is None:
        nodo = self.raiz

    return self._altura_recursiva(nodo)

def _altura_recursiva(self, nodo):
    if nodo is None:
        return -1

    return 1 + max(
        self._altura_recursiva(nodo.izquierdo),
        self._altura_recursiva(nodo.derecho)
    )
````

# Pruebas de la tarea

En esta sección se crea un árbol de prueba y se ejecutan todos los métodos implementados.

Las pruebas de eliminación se realizan en este orden:

1. Se elimina `10`, que inicialmente es una hoja.
2. Se elimina `20`, que después de eliminar `10` queda con un solo hijo: `25`.
3. Se elimina `30`, que tiene dos hijos.

También se muestra:

- Recorrido inorden.
- Recorrido BFS.
- Suma total de valores.
- Validación de nodos internos.
- Árbol balanceado.
- Comparación de alturas.

````python
arbol = ArbolBinarioBusqueda()

valores = [50, 30, 70, 20, 40, 60, 80, 10, 25, 35, 45]

for valor in valores:
    arbol.insertar(valor)

print("ÁRBOL ORIGINAL")
arbol.mostrar_arbol()

print("\nRecorrido inorden:", arbol.inorden())
print("Recorrido por niveles:", arbol.recorrido_por_niveles())
print("Suma de elementos:", arbol.sumar_elementos())
print("¿Todos los nodos internos tienen dos hijos?:", arbol.es_completo())
print("Altura del árbol original:", arbol.altura())

arbol.eliminar(10)
print("\nDespués de eliminar hoja 10:", arbol.inorden())

arbol.eliminar(20)
print("Después de eliminar nodo con un hijo 20:", arbol.inorden())

arbol.eliminar(30)
print("Después de eliminar nodo con dos hijos 30:", arbol.inorden())

arbol_balanceado = arbol.balancear_arbol()

print("\nÁRBOL BALANCEADO")
arbol_balanceado.mostrar_arbol()

print("\nInorden del árbol balanceado:", arbol_balanceado.inorden())
print("Altura del árbol balanceado:", arbol_balanceado.altura())
````
